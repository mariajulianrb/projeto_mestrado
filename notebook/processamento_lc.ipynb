## Subtração do fundo na curva de luz

```python
import sys
import os
sys.path.append('../src') 
from processamento import subtrair_fundo

ano = "2022"
binagem = "b1h"
energia = "15-50kev"

pasta_destino = "../data/processados/meus_testes/"

raios = {
    '2022': {'src': 18.984525, 'bkg': 38.948665},
    '2017': {'src': 20.500000, 'bkg': 40.000000}
}

os.makedirs(pasta_destino, exist_ok=True)

arq_fonte = f"../data/Dados {ano}/{binagem}/{energia}/nu30701024002A01_sr.lc"
arq_fundo = f"../data/Dados {ano}/{binagem}/{energia}/nu30701024002A01_bk.lc"
arq_saida = f"{pasta_destino}lc_sub_{ano}_{binagem}_{energia}.lc"

subtrair_fundo(arq_fonte, arq_fundo, raios[ano]['src'], raios[ano]['bkg'], arq_saida)
```

```python
import matplotlib.pyplot as plt
from astropy.table import Table

# Carrega o ficheiro processado que acabou de salvar
tabela = Table.read(arq_saida, hdu=1)
df = tabela.to_pandas()

# Conversão do tempo da missão NuSTAR para MJD
mjd_base = 55197.00076601852
df['MJD'] = mjd_base + (df['TIME'] / 86400.0)

# Gráfico com barras de erro incluídas
plt.figure(figsize=(9, 4.5))
plt.errorbar(df['MJD'], df['RATE'], yerr=df['ERROR'], fmt='b.', alpha=0.5, markersize=2, elinewidth=0.5)

plt.ticklabel_format(useOffset=False, style='plain')
plt.xlabel('Tempo (MJD)')
plt.ylabel('Taxa de Contagens Líquida (count/s)')
plt.title(f'Curva de Luz LS5039 - Corrigida ({energia})')
plt.grid(True, ls=':', alpha=0.6)

plt.tight_layout()
plt.show()
```
