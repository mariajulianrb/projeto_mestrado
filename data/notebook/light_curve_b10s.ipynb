```python
import matplotlib.pyplot as plt
import pandas as pd
from astropy.table import Table

caminho = 'data/dados 10s/3-10kev/nu30201034002A01_sr.lc'
tabela = Table.read(caminho, hdu=1)
df = tabela.to_pandas()

df.columns = [col.capitalize() for col in df.columns]

df['Time'] = pd.to_numeric(df['Time'], errors='coerce')
df['Rate'] = pd.to_numeric(df['Rate'], errors='coerce')

df = df.dropna(subset=['Time', 'Rate'])
df = df[df['Rate'] > 0]

plt.figure(figsize=(10, 5))
plt.plot(df['Time'], df['Rate'], marker='.', linestyle='', color='blue')

plt.xlabel('Tempo (s)')
plt.ylabel('Taxa de Contagens (count/s)')
plt.title('Curva de Luz (3-10keV) com binagem de 10s')
plt.ylim(0,10.0)


plt.tight_layout()
plt.show()
```
