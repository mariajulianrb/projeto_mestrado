## As curvas de luz foram gerada com bins = 1s

```python
import matplotlib.pyplot as plt
import pandas as pd
from astropy.table import Table

caminho = '../../data/bins 1s/nu30201034002A01_sr_3-10kev_b1s.lc'
tabela = Table.read(caminho, hdu=1)
df = tabela.to_pandas()

df.columns = [col.capitalize() for col in df.columns]

df['Time'] = pd.to_numeric(df['Time'], errors='coerce')
df['Rate'] = pd.to_numeric(df['Rate'], errors='coerce')

df = df.dropna(subset=['Time', 'Rate'])
df = df[df['Rate'] > 0]

plt.figure(figsize=(10, 5))
plt.plot(df['Time'], df['Rate'], marker='.', linestyle='', color='blue')

plt.xlabel('Tempo (s)')
plt.ylabel('Taxa de Contagens (count/s)')
plt.title('Curva de Luz (3-10keV) com binagem de 1s')
plt.ylim(0,10.0)


plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from astropy.table import Table
from astropy.timeseries import LombScargle

caminho = 'data/bins 1s/nu30201034002A01_sr_30-70kev_b1s.lc'
tabela = Table.read(caminho, hdu=1)

time = np.array(tabela['TIME'])
rate = np.array(tabela['RATE'])
err = np.array(tabela['ERROR']) if 'ERROR' in tabela.colnames else None

mask = (rate > 0) & ~np.isnan(rate)

t_sec = time[mask] 

rate_clean = rate[mask] - np.mean(rate[mask])

ls = LombScargle(t_sec, rate_clean, err[mask] if err is not None else None)

f_min_hz, f_max_hz = 0.001, 0.5
freq_hz, power = ls.autopower(
    minimum_frequency=f_min_hz, 
    maximum_frequency=f_max_hz, 
    samples_per_peak=10
)

idx_pico = np.argmax(power)
best_freq = freq_hz[idx_pico]
best_p = 1 / best_freq 

print(f'Melhor frequência: {best_freq:.5f} Hz')
print(f'Período correspondente: {best_p:.2f} segundos')

plt.figure(figsize=(9, 4.5))
plt.plot(freq_hz, power, 'k-', lw=1.2)
plt.axvline(
    best_freq,
    color='r',
    ls='--',
    label=f'f = {best_freq:.4f} Hz (P = {best_p:.1f} s)',
)
plt.xlabel('Frequência (Hz)')
plt.ylabel('Potência')
plt.title('Periodograma Lomb-Scargle - Alta Frequência (30-70 keV)')
plt.xlim(f_min_hz, f_max_hz)
plt.legend()
plt.grid(True, ls=':', alpha=0.6)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
from astropy.table import Table
from astropy.timeseries import LombScargle

caminho = 'data/bins 1h/nu30201034002A01_sr10-30keV_b1h.lc'
tabela = Table.read(caminho, hdu=1)

time = np.array(tabela['TIME'])
rate = np.array(tabela['RATE'])
err = np.array(tabela['ERROR']) if 'ERROR' in tabela.colnames else None

mask = (rate > 0) & ~np.isnan(rate)

t_dias = time[mask] / 86400.0
rate_clean = rate[mask] - np.mean(rate[mask]) 

min_p, max_p = 0.1, 5.0
freq_min, freq_max = 1 / max_p, 1 / min_p

ls = LombScargle(t_dias, rate_clean, err[mask] if err is not None else None)
freq, power = ls.autopower(
    minimum_frequency=freq_min, maximum_frequency=freq_max, samples_per_peak=10
)

idx_pico = np.argmax(power)
best_freq = freq[idx_pico]
best_p = 1 / best_freq

p_esperado = 3.90603  
f_esperado = 1 / p_esperado

print(f'Melhor frequência: {best_freq:.4f} dias⁻¹ (P = {best_p:.2f} dias)')
print(f'Frequência esperada: {f_esperado:.4f} dias⁻¹ (P = {p_esperado} dias)')

plt.figure(figsize=(8, 4))
plt.plot(freq, power, 'b-', lw=1.5)
plt.axvline(
    f_esperado,
    color='r',
    ls='--',
    label=f'Esperado (P = {p_esperado} d)',
)
plt.axvline(
    best_freq, color='k', ls=':', label=f'Pico (P = {best_p:.2f} d)'
)
plt.xlabel('Frequência (dias⁻¹)')
plt.ylabel('Potência')
plt.title('Periodograma Lomb-Scargle - Escala Orbital')
plt.legend()
plt.grid(True, ls=':', alpha=0.6)
plt.tight_layout()
plt.show()
```
