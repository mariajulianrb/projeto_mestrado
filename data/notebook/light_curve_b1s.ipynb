## As curvas de luz foram gerada com bins = 1s

```python
import matplotlib.pyplot as plt
import pandas as pd
from astropy.table import Table

tabela = Table.read('nu30201034002A01_sr_3-10kev_b1s.lc', hdu=1)
df = tabela.to_pandas()

df.columns = [col.capitalize() for col in df.columns]

df['Time'] = pd.to_numeric(df['Time'], errors='coerce')
df['Rate'] = pd.to_numeric(df['Rate'], errors='coerce')

df = df.dropna(subset=['Time', 'Rate'])
df = df[df['Rate'] > 0]

plt.figure(figsize=(10, 5))
plt.plot(df['Time'], df['Rate'], marker='.', linestyle='', color='blue')

plt.xlabel('Tempo (s)')
plt.ylabel('Taxa de Contagens (count/s)')
plt.title('Curva de Luz (3-10keV)')


plt.tight_layout()
plt.show()
```

## Leitura do Header

```python
from astropy.io import fits

hdr = fits.getheader('nu30201034002A01_sr_3-10kev_b1s.lc', hdu=1)

chaves = [
    'OBJECT',
    'TELESCOP',
    'INSTRUME',
    'OBS_ID',
    'DATE-OBS',
    'DATE-END',
    'MJDREFI',
    'MJDREFF',
    'TSTART',
    'TSTOP',
    'ONTIME',
]

for chave in chaves:
    if chave in hdr:
        print(f'{chave:10s} = {hdr[chave]} \t/ {hdr.comments[chave]}')
```

## conversão para MJD

```python
import matplotlib.pyplot as plt
from astropy.table import Table

df = Table.read('nu30201034002A01_sr_3-10kev_b1s.lc', hdu=1).to_pandas()

mjd_base = 55197.00076601852
df['MJD'] = mjd_base + (df['TIME'] / 60 / 60 / 24)

plt.plot(df['MJD'], df['RATE'], 'b.')
plt.ticklabel_format(useOffset=False, style='plain')
plt.xlabel('Tempo (MJD)')
plt.ylabel('Taxa de Contagens (count/s)')
plt.title('Curva de Luz (3-10 keV)')
plt.show()
```
